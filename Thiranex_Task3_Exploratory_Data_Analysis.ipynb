## Roller Coaster Database

This project performs exploratory data analysis on a roller coaster dataset to understand its structure, characteristics, patterns, and relationships between different features.

The analysis follows the EDA workflow demonstrated in the Thiranex tutorial, including data understanding, data preparation, feature exploration, feature relationships, and asking questions about the data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# 2. Feature Understanding

```python
plt.figure(figsize=(12, 6))

sns.histplot(df["year_introduced"], bins=30)

plt.title("Distribution of Roller Coasters by Year Introduced")
plt.xlabel("Year Introduced")
plt.ylabel("Number of Roller Coasters")

plt.show()
```

```python
df["Type_Main"].value_counts()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Type_Main")

plt.title("Number of Roller Coasters by Type")
plt.xlabel("Coaster Type")
plt.ylabel("Number of Roller Coasters")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(df["speed_mph"].dropna(), bins=30)

plt.title("Distribution of Roller Coaster Speeds")
plt.xlabel("Speed (mph)")
plt.ylabel("Number of Roller Coasters")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(df["height_value"].dropna(), bins=30)

plt.title("Distribution of Roller Coaster Heights")
plt.xlabel("Height")
plt.ylabel("Number of Roller Coasters")

plt.show()
```

# 3. Feature Relationships

```python
sns.scatterplot(
    data=df,
    x="height_value",
    y="speed_mph"
)

plt.title("Roller Coaster Height vs Speed")
plt.xlabel("Height")
plt.ylabel("Speed (mph)")

plt.show()
```

```python
df[["height_value", "speed_mph"]].corr()
```

# 4. Asking Questions About the Data

```python
corr = df[[
    "speed_mph",
    "height_value",
    "Inversions_clean",
    "Gforce_clean",
    "year_introduced"
]].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Between Roller Coaster Features")

plt.show()
```

```python
df.groupby("Type_Main")["speed_mph"].mean()
```

```python
avg_speed = df.groupby("Type_Main")["speed_mph"].mean().reset_index()

plt.figure(figsize=(8, 5))

sns.barplot(data=avg_speed, x="Type_Main", y="speed_mph")

plt.title("Average Roller Coaster Speed by Type")
plt.xlabel("Coaster Type")
plt.ylabel("Average Speed (mph)")

plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="Inversions_clean",
    y="Gforce_clean"
)

plt.title("Inversions vs G-Force")
plt.xlabel("Number of Inversions")
plt.ylabel("G-Force")

plt.show()
```

```python
df[["Inversions_clean", "Gforce_clean"]].corr()
```

```python
df["Manufacturer"].value_counts().head(10)
```

```python
top_manufacturers = df["Manufacturer"].value_counts().head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_manufacturers.values,
    y=top_manufacturers.index
)

plt.title("Top 10 Roller Coaster Manufacturers")
plt.xlabel("Number of Roller Coasters")
plt.ylabel("Manufacturer")

plt.show()
```

```python
df[["coaster_name", "speed_mph", "Type_Main", "Manufacturer"]].sort_values(
    "speed_mph",
    ascending=False
).head(10)
```

# Final Insights

## Key Findings

- The dataset contains 1,087 roller coasters across 56 columns.
- Steel coasters are the most common type in the dataset, followed by wooden and other types.
- The distribution of roller coasters by introduction year shows how coaster development has changed over time.
- Roller coaster speeds vary considerably, with most recorded speeds concentrated below the highest-speed outliers.
- There is a fairly strong positive correlation between coaster height and speed (r ≈ 0.70), meaning taller coasters in this dataset tend to also have higher speeds.
- The correlation between inversions and G-force is moderate (r ≈ 0.35), suggesting that coasters with more inversions tend to have somewhat higher G-forces.
- Vekoma, Bolliger & Mabillard, and Intamin are the three most represented manufacturers in the dataset.
- The fastest coaster in the dataset is Formula Rossa at 149.1 mph.
- The fastest 10 coasters identified in the analysis are all classified as steel coasters.

## Conclusion

The exploratory analysis reveals several interesting patterns in the roller coaster dataset. Speed and height show a relatively strong positive relationship, while inversions and G-force have a weaker positive relationship. Steel coasters dominate the dataset and also appear prominently among the fastest recorded coasters. The analysis demonstrates how statistical summaries, visualizations, and correlation analysis can be used to understand patterns and relationships within a real-world dataset.
